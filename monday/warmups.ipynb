## 6. Images are numbers

Here's the secret the whole week rests on: **a picture is just a grid of numbers.** Each little dot (a
**pixel**) is a number for how bright it is — 0 is black, up to 255 is white. Your model never sees "a mug";
it sees a grid of numbers and learns the patterns. Let's make one by hand.


```python
# A tiny 8x8 picture. Each number is a pixel's brightness, 0 (dark) to 9 (bright).
# (Real photos go 0-255; we use 0-9 here so it fits on screen.)
picture = [
    [0, 0, 9, 9, 9, 9, 0, 0],
    [0, 9, 0, 0, 0, 0, 9, 0],
    [9, 0, 9, 0, 0, 9, 0, 9],
    [9, 0, 0, 0, 0, 0, 0, 9],
    [9, 0, 9, 0, 0, 9, 0, 9],
    [9, 0, 0, 9, 9, 0, 0, 9],
    [0, 9, 0, 0, 0, 0, 9, 0],
    [0, 0, 9, 9, 9, 9, 0, 0],
]

print("The numbers:\n")
for row in picture:
    print(" ".join(str(v) for v in row))
```

Those are the numbers. Now let's **see the picture** — we swap each number for a shade,
from a space (dark) up to `@` (bright). Same grid, drawn instead of listed.


```python
shades = " .:-=+*#%@"   # 10 shades, dark -> light; shades[0]=' ', shades[9]='@'

print("The picture:\n")
for row in picture:
    print("".join(shades[v] for v in row))
```

See the face? That's the whole idea: **numbers on the left, a picture on the right — they're the same thing.**
