## 4. Functions

A **function** is a reusable mini-program. You give it inputs, it gives back a result.
Here's one that turns a model's confidence score into a friendly message.


```python
def describe_confidence(score):
    if score >= 0.90:
        return "very sure"
    elif score >= 0.60:
        return "fairly sure"
    else:
        return "not sure"

print(describe_confidence(0.99))
print(describe_confidence(0.72))
print(describe_confidence(0.30))
```

## 5. A peek at your training data

On Wednesday you'll train your model from the photos you collect today. The photos live in folders,
**one folder per class**. The cell below counts how many photos you have in each class and draws a little bar.
Right now you'll probably have zero of everything — that's exactly today's job.


```python
from pathlib import Path

# Use the same class names you chose above:
classes = ["object_1", "object_2", "object_3", "none"]

data_dir = Path("data/train")
image_types = {".jpg", ".jpeg", ".png"}

print("Photos per class:\n")
for c in classes:
    folder = data_dir / c
    if folder.exists():
        count = len([p for p in folder.iterdir() if p.suffix.lower() in image_types])
    else:
        count = 0
    bar = "#" * count
    print(f"  {c:12s} {count:3d}  {bar}")

print("\nGoal: about 40-60 photos per class, balanced across classes.")
print("If everything shows 0, that's expected before you start shooting.")
```
